###     Basics of grouping

```python
# groupby method is groups the data by different catogories. the data is grouped baswed on one or several variable qnd 
# analysis is performed on the individual group 
```

```python
df['drive-wheels'].unique().tolist()
# a vehicle drive-wheels is the wheel tire assembly that actually pushes or pulls the veh down the road
```

```python
#which type of drive wheels is most valuable
df_group_one = df[['drive-wheels', 'body-style', 'price']]
df_group_one = df_group_one.groupby(['drive-wheels'], as_index=False).mean(numeric_only=True)
df_group_one
```

```python
 #which type of drive wheels is most valuable
df_gptset = df[['drive-wheels', 'body-style', 'price']]
grouped_test = df_gptset.groupby(['drive-wheels', 'body-style'], as_index=False).mean()
grouped_test
```

```python
# pivot table means creating a spreadsheet-style table that summarizes data by aggregating values based on specified rows and columns
pivot_table = df.pivot_table(values='price', index='drive-wheels', columns='body-style')
pivot_table
```

```python
pivot_table= pivot_table.fillna(0)
pivot_table
```
